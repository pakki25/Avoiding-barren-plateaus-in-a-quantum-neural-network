Step 3 : Define a cost function and Calculate gradients

```python
import numpy as np
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

np.random.seed(42)

# Function to create the QNN
def create_qnn(n, params):
    qc = QuantumCircuit(n)

    # First rotation layer
    for q in range(n):
        qc.ry(params[0, q], q)

    # First entanglement layer
    for q in range(n - 1):
        qc.cx(q, q + 1)

    # Second rotation layer
    for q in range(n):
        qc.ry(params[1, q], q)

    # Second entanglement layer
    for q in range(n - 1):
        qc.cx(q, q + 1)

    # Third rotation layer
    for q in range(n):
        qc.ry(params[2, q], q)

    return qc


# Function to calculate the cost
def calculate_cost(n, params):
    qc = create_qnn(n, params)

    # Simulate the quantum circuit
    state = Statevector.from_instruction(qc)

    # Probability of the desired output |00...0>
    probability_zero = abs(state.data[0]) ** 2

    # Cost function
    return 1 - probability_zero


# Function to calculate the gradient
def calculate_gradient(n, params):
    plus = params.copy()
    minus = params.copy()

    # Shift the first rotation parameter
    plus[0, 0] += np.pi / 2
    minus[0, 0] -= np.pi / 2

    # Parameter-shift rule
    cost_plus = calculate_cost(n, plus)
    cost_minus = calculate_cost(n, minus)

    gradient = (cost_plus - cost_minus) / 2

    return gradient


# Calculate gradients for 2 to 10 qubits
for n in range(2, 11):

    # Generate random parameters for 3 layers
    params = np.random.uniform(0, 2 * np.pi, size=(3, n))

    # Calculate cost and gradient
    cost = calculate_cost(n, params)
    gradient = calculate_gradient(n, params)

    print(f"Number of qubits: {n}")
    print(f"Cost: {cost:.6f}")
    print(f"Gradient: {gradient:.8f}")
    print("-" * 35)

```

Step 4 : Repeat for multiple random parameter settings and calculate gradient variance

```python
import numpy as np
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

np.random.seed(42)

# Create the QNN circuit
def create_qnn(n, params):
    qc = QuantumCircuit(n)

    # First rotation layer
    for q in range(n):
        qc.ry(params[0, q], q)

    # First entanglement layer
    for q in range(n - 1):
        qc.cx(q, q + 1)

    # Second rotation layer
    for q in range(n):
        qc.ry(params[1, q], q)

    # Second entanglement layer
    for q in range(n - 1):
        qc.cx(q, q + 1)

    # Third rotation layer
    for q in range(n):
        qc.ry(params[2, q], q)

    return qc


# Calculate the cost function
def calculate_cost(n, params):
    qc = create_qnn(n, params)
    state = Statevector.from_instruction(qc)

    # Probability of the desired output |00...0>
    probability_zero = abs(state.data[0]) ** 2

    return 1 - probability_zero


# Calculate the gradient using parameter-shift rule
def calculate_gradient(n, params):
    plus = params.copy()
    minus = params.copy()

    plus[0, 0] += np.pi / 2
    minus[0, 0] -= np.pi / 2

    cost_plus = calculate_cost(n, plus)
    cost_minus = calculate_cost(n, minus)

    return (cost_plus - cost_minus) / 2


# Experiment settings
qubit_values = range(2, 11)
trials = 100

gradient_variances = []

# Repeat experiment for 2 to 10 qubits
for n in qubit_values:

    gradients = []

    for trial in range(trials):

        # Generate new random parameters for each trial
        params = np.random.uniform(
            0, 2 * np.pi, size=(3, n)
        )

        # Calculate gradient
        gradient = calculate_gradient(n, params)
        gradients.append(gradient)

    # Calculate gradient variance
    variance = np.var(gradients)
    gradient_variances.append(variance)

    print(f"Number of qubits: {n}")
    print(f"Average gradient: {np.mean(gradients):.8f}")
    print(f"Gradient variance: {variance:.10f}")
    print("-" * 40)

print("Gradient variances for 2 to 10 qubits:")
print(gradient_variances)

```

```python
import numpy as np

def calculate_all_gradients(n, params):
    gradients = np.zeros_like(params)
    shift = np.pi / 2

    for i in range(params.shape[0]):
        for j in range(params.shape[1]):
            params_plus = params.copy()
            params_minus = params.copy()

            params_plus[i, j] += shift
            params_minus[i, j] -= shift

            cost_plus = calculate_cost(n, params_plus)
            cost_minus = calculate_cost(n, params_minus)

            gradients[i, j] = (cost_plus - cost_minus) / 2

    return gradients
```

Step 1: Create the training function

```python
def train_qnn(n, initialization, iterations=100, learning_rate=0.1):

    # Initialize the parameters
    if initialization == "random":
        params = np.random.uniform(0, 2*np.pi, size=(3, n))

    elif initialization == "near_identity":
        params = np.random.uniform(-0.1, 0.1, size=(3, n))

    # Store the cost at each iteration
    cost_history = []

    for step in range(iterations):

        # Calculate the current cost
        cost = calculate_cost(n, params)
        cost_history.append(cost)

        # Calculate gradients for all parameters
        gradients = calculate_all_gradients(n, params)

        # Update all parameters using gradient descent
        params = params - learning_rate * gradients

    return cost_history, params
```

Step 2: Train using both initialization methods

```python
n = 4

# Train with random initialization
random_costs, random_final_params = train_qnn(
    n,
    initialization="random",
    iterations=100,
    learning_rate=0.1
)

# Train with near-identity initialization
near_identity_costs, near_identity_final_params = train_qnn(
    n,
    initialization="near_identity",
    iterations=100,
    learning_rate=0.1
)

print("Random initialization")
print("Initial cost:", random_costs[0])
print("Final recorded cost:", random_costs[-1])

print("\nNear-identity initialization")
print("Initial cost:", near_identity_costs[0])
print("Final recorded cost:", near_identity_costs[-1])
```

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

simulator = AerSimulator()


def calculate_cost_aer(n, params, shots=1024):

    qc = QuantumCircuit(n)

    # Three layers of rotation gates
    for layer in range(3):

        for qubit in range(n):
            qc.ry(params[layer, qubit], qubit)

        # Entangle neighboring qubits
        for qubit in range(n - 1):
            qc.cx(qubit, qubit + 1)

    # Measure all qubits
    qc.measure_all()

    # Run the circuit
    result = simulator.run(
        qc,
        shots=shots
    ).result()

    counts = result.get_counts()

    # Number of all-zero measurement outcomes
    zero_state = "0" * n
    zero_count = counts.get(zero_state, 0)

    probability = zero_count / shots

    # Cost function
    cost = 1 - probability

    return cost
```

```python
import numpy as np

# Test with 4 qubits
n = 4

# Random parameters for 3 layers
params = np.random.uniform(
    0, 2 * np.pi,
    size=(3, n)
)

cost = calculate_cost_aer(n, params, shots=1024)

print("Number of qubits:", n)
print("Estimated QNN cost:", cost)
```

```python
from qiskit.quantum_info import Statevector

# Calculate exact cost using the ideal statevector
def calculate_cost_ideal(n, params):

    qc = QuantumCircuit(n)

    for layer in range(3):

        for qubit in range(n):
            qc.ry(params[layer, qubit], qubit)

        for qubit in range(n - 1):
            qc.cx(qubit, qubit + 1)

    state = Statevector.from_instruction(qc)

    probability = abs(state.data[0])**2

    return 1 - probability


# Compare both simulators using the same parameters
ideal_cost = calculate_cost_ideal(n, params)

aer_cost = calculate_cost_aer(
    n, params, shots=1024
)

print("Ideal simulator cost:", ideal_cost)
print("Aer simulator cost:", aer_cost)
print("Absolute difference:", abs(ideal_cost - aer_cost))
```

Run 20 Aer trials

```python
import numpy as np

# Use the same four-qubit parameters throughout the experiment
np.random.seed(42)

n = 4

params = np.random.uniform(
    0, 2 * np.pi,
    size=(3, n)
)

shots = 1024
num_trials = 20

# Exact reference cost
ideal_cost = calculate_cost_ideal(n, params)

# Run Aer 20 times with identical parameters
aer_costs = []

for trial in range(num_trials):
    cost = calculate_cost_aer(
        n, params, shots=shots
    )

    aer_costs.append(cost)

aer_costs = np.array(aer_costs)

print("Number of trials:", num_trials)
print("Shots per trial:", shots)
print("Ideal statevector cost:", ideal_cost)
print("Aer mean cost:", np.mean(aer_costs))
print("Aer standard deviation:", np.std(aer_costs, ddof=1))
print("Mean absolute error from ideal:",
      np.mean(np.abs(aer_costs - ideal_cost)))

print("\nIndividual Aer costs:")
print(aer_costs)
```
